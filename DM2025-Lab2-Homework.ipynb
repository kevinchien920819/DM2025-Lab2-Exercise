## 1. Preprocessing Steps

```python
import pandas as pd
import json
import re
import emoji
import contractions
import plotly.express as px
from sentence_transformers import SentenceTransformer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import f1_score

DATA_DIR = 'kaggle_dataset'
emotion = pd.read_csv(f'{DATA_DIR}/emotion.csv')
data_identification = pd.read_csv(f'{DATA_DIR}/data_identification.csv')
# load json raw data 
with open(f'{DATA_DIR}/final_posts.json', 'r') as f:
    raw_posts = json.load(f)
records = []
for item in raw_posts:
    try:
        post = item["root"]["_source"]["post"]
        pid = post["post_id"]
        text = post.get("text", "")
        records.append((pid, text))
    except Exception:
        continue
posts_df = pd.DataFrame(records, columns=["id", "text"])
data_merged = data_identification.merge(posts_df, on="id", how="left")

# Build train and test sets
train_df = data_merged[data_merged["split"] == "train"].merge(emotion, on="id", how="left")
test_df = data_merged[data_merged["split"] == "test"].copy()
```

## 2. Feature Engineering Steps

```python
mention_pattern = re.compile(r"@\w+")
hashtag_pattern = re.compile(r"#(\w+)")
url_pattern = re.compile(r"http\S+|www\.\S+")
emoji_pattern = re.compile("[\U00010000-\U0010ffff]", flags=re.UNICODE)

# define the clean text method 
def clean_text(text: str) -> str:

    text = text.lower()

    # remove mentions
    text = mention_pattern.sub(" ", text)

    # keep hashtag word
    text = hashtag_pattern.sub(r"\1", text)

    # remove URLs
    text = url_pattern.sub(" ", text)

    # convert emoji to text
    text = emoji.demojize(text, delimiters=(" ", " "))

    # expand contractions
    text = contractions.fix(text)

    # reduce repeated letters
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)

    # remove html tags
    text = re.sub(r"<.*?>", " ", text)

    # remove non-alphanumeric except some punctuation
    text = re.sub(r"[^a-z0-9!?'\s]", " ", text)

    # clean extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text

train_texts = train_df["text"].astype(str).apply(clean_text)
test_texts = test_df["text"].astype(str).apply(clean_text)

train_df["text"] = train_texts
test_df["text"] = test_texts
# fill NA
train_df["text"] = train_df["text"].astype(str).fillna("NaN")
test_df["text"] = test_df["text"].astype(str).fillna("NaN")
# save cleaned dataset 
train_df.to_csv(f'{DATA_DIR}/train_df.csv',index=None)
test_df.to_csv(f'{DATA_DIR}/test_df.csv',index=None)
```

## Bert base pre-train

```python
import torch
import pandas as pd
from torch.utils.data import Dataset

# Training set
class TrainDataset(Dataset):
    def __init__(self, df, tokenizer, max_len=128):
        self.df = df
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        text = str(row["text"])

        encoded = self.tokenizer(
            text,
            padding="max_length",
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )

        return {
            "input_ids": encoded["input_ids"].squeeze(),
            "attention_mask": encoded["attention_mask"].squeeze(),
            "labels": torch.tensor(row["label_id"], dtype=torch.long)
        }
# Testing set
class TestDataset(Dataset):
    def __init__(self, df, tokenizer, max_len=128):
        self.df = df
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        text = str(row["text"])

        encoded = self.tokenizer(
            text,
            padding="max_length",
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )

        return {
            "input_ids": encoded["input_ids"].squeeze(),
            "attention_mask": encoded["attention_mask"].squeeze()
        }
```

```python
import pandas as pd
import torch
import os
from torch import nn
from torch.utils.data import DataLoader
from torch.optim import AdamW
from tqdm.auto import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification



# -------------------------------------------------------
# Training function
# -------------------------------------------------------
def train_one_epoch(model, dataloader):
    model.train()
    total_loss = 0

    pbar = tqdm(dataloader, desc="Training", leave=False)

    for batch in pbar:
        optimizer.zero_grad()

        outputs = model(
            batch["input_ids"].to(device),
            attention_mask=batch["attention_mask"].to(device)
        )

        logits = outputs.logits
        loss = criterion(logits, batch["labels"].to(device))

        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        pbar.set_postfix({"batch_loss": loss.item()})

    return total_loss / len(dataloader)


# -------------------------------------------------------
# Prediction function
# -------------------------------------------------------
def predict(model, dataloader):
    model.eval()
    pred_ids = []

    with torch.no_grad():
        pbar = tqdm(dataloader, desc="Predicting")
        for batch in pbar:
            outputs = model(
                batch["input_ids"].to(device),
                attention_mask=batch["attention_mask"].to(device)
            )
            logits = outputs.logits
            preds = torch.argmax(logits, dim=1).cpu().tolist()
            pred_ids.extend(preds)

    return pred_ids


# -------------------------------------------------------
# Main
# -------------------------------------------------------
if __name__ == "__main__":

    DATA_DIR = "kaggle_dataset"


    # -------------------------------------------------------
    # load data
    # -------------------------------------------------------
    train_df = pd.read_csv(f"{DATA_DIR}/train_df.csv")
    test_df = pd.read_csv(f"{DATA_DIR}/test_df.csv")

    # 建 label-id 映射
    label2id = {l: i for i, l in enumerate(train_df["emotion"].unique())}
    id2label = {i: l for l, i in label2id.items()}

    train_df["label_id"] = train_df["emotion"].map(label2id)
    tokenizer = AutoTokenizer.from_pretrained("bhadresh-savani/bert-base-uncased-emotion")

    # -------------------------------------------------------
    # Dataset / DataLoader
    # -------------------------------------------------------
    train_loader = DataLoader(
        TrainDataset(train_df, tokenizer),
        batch_size=16,
        shuffle=True
    )

    test_loader = DataLoader(
        TestDataset(test_df.fillna(""), tokenizer),
        batch_size=32
    )


    # -------------------------------------------------------
    # Model / Optimizer / Loss
    # -------------------------------------------------------
    device = "cuda" if torch.cuda.is_available() else "cpu"

    model = AutoModelForSequenceClassification.from_pretrained(
        "bhadresh-savani/bert-base-uncased-emotion"
    ).to(device)

    criterion = nn.CrossEntropyLoss()
    optimizer = AdamW(model.parameters(), lr=2e-5)


    # ---- Training ----
    EPOCHS = 1
    for ep in range(EPOCHS):
        loss = train_one_epoch(model, train_loader)
        print(f"Epoch {ep+1}/{EPOCHS} - Loss: {loss:.4f}")

    # ---- Predict ----
    pred_ids = predict(model, test_loader)

    test_df["emotion"] = [id2label[i] for i in pred_ids]
    saving_path = "kaggle_dataset/predict"
    os.makedirs(saving_path, exist_ok=True)

    output_file = os.path.join(saving_path, f"bert_submission_{EPOCHS}epoch.csv")

    test_df[["id", "emotion"]].to_csv(output_file, index=False)

    print(f"Saved result in:{output_file}")
```

## LLM embedding 

```python
import pandas as pd
import json
import re
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split

# -----------------------
# 1. Load data
# -----------------------
emotion = pd.read_csv('kaggle_dataset/emotion.csv')
data_identification = pd.read_csv('kaggle_dataset/data_identification.csv')

# Load json posts
with open('kaggle_dataset/final_posts.json', 'r') as f:
    raw_posts = json.load(f)

records = []
for item in raw_posts:
    try:
        post = item["root"]["_source"]["post"]
        pid = post["post_id"]
        text = post.get("text", "")
        records.append((pid, text))
    except:
        pass

posts_df = pd.DataFrame(records, columns=["id", "text"])

# Merge with split
merged = data_identification.merge(posts_df, on="id", how="left")

# Build train & test
train_df = merged[merged["split"] == "train"].merge(emotion, on="id", how="left")
test_df  = merged[merged["split"] == "test"]

train_df = train_df.dropna(subset=["text", "emotion"])
test_df = test_df.dropna(subset=["text"])

X_text = train_df["text"].astype(str)
y = train_df["emotion"].astype(str)
X_test_text = test_df["text"].astype(str)

# -----------------------
# 2. Basic text cleaning
# -----------------------
def clean_text(t):
    t = t.lower()
    t = re.sub(r"http\S+|www\.\S+", " ", t)
    t = re.sub(r"@\w+", " ", t)       # remove mentions
    t = re.sub(r"#(\w+)", r"\1", t)   # keep hashtag word
    t = re.sub(r"[^a-zA-Z\s]", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

X_clean = X_text.apply(clean_text)
X_test_clean = X_test_text.apply(clean_text)

# -----------------------
# 3. LLM Embedding
# -----------------------
print("Loading sentence-transformer model...")
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

print("Encoding train texts...")
X_emb = model.encode(X_clean.tolist(), show_progress_bar=True)

print("Encoding test texts...")
X_test_emb = model.encode(X_test_clean.tolist(), show_progress_bar=True)

# -----------------------
# 4. Train classifier
# -----------------------
X_train, X_val, y_train, y_val = train_test_split(X_emb, y, test_size=0.2, stratify=y)

clf = LogisticRegression(max_iter=2000, class_weight="balanced")
clf.fit(X_train, y_train)

val_pred = clf.predict(X_val)
f1 = f1_score(y_val, val_pred, average="macro")
print("Validation Macro-F1:", f1)

# Retrain on full data
clf.fit(X_emb, y)

# -----------------------
# 5. Predict test set
# -----------------------
test_pred = clf.predict(X_test_emb)

# -----------------------
# 6. Build submission
# -----------------------
sub = pd.DataFrame({"id": test_df["id"].values,
                    "emotion": test_pred})

sub.to_csv("llm_embedding_submission.csv", index=False)
sub.head()
```
